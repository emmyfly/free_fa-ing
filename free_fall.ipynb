```python
# first project
#free fall simulation


g = 9.8 # gravity in meters per seconds square
start_height = 25 # meters
t = 0 # time in seconds
height = start_height

height = height - 0.5 *g*t*t

while height >0:
    print(round(t,1), round(height,1))
    t = t+0.1
    height = start_height - 0.5*g*t*t
```

```python
import matplotlib.pyplot as plt
g = 9.8
t = 0
start_height = 25

height = start_height 

time=[]   # empty list for the results
heights =[]

while height > 0:
      time.append(t)
      heights.append(height)
      t = t + 0.1
      height = start_height - 0.5*g*t*t

plt.plot(time,heights)
plt.xlabel("time(s)")
plt.ylabel("height(m)")
plt.title(f"Ball dropped from { 25}m")
plt.show()
```
